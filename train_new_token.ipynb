```python
def get_training_corpus():
    dataset = raw_dataset["train"]

    for start_idx in range(0, len(dataset), 1000):
        samples = dataset[start_idx : start_idx + 1000]
        yield samples["whole_func_string"]
```

```python
from transformers import AutoTokenizer

training_corpus = get_training_corpus()
old_tokenizer = AutoTokenizer.from_pretrained("gpt2")
new_tokenizer=old_tokenizer.train_new_from_iterator(training_corpus, 52000)
```

```python
print(new_tokenizer.vocab_size)
```

```python
vocab = new_tokenizer.get_vocab()

print(vocab)
```

```python
for token, token_id in list(vocab.items())[:50]:
    print(token, "→", token_id)
```

```python
code = """
def calculate_average(numbers):
    return sum(numbers) / len(numbers)
"""

tokens = new_tokenizer.tokenize(code)

print(tokens)
```

```python
print(new_tokenizer(code))
```

```python
print(new_tokenizer.encode(code))
```
